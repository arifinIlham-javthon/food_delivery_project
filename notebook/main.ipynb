--- Sprint 1

```python
import pandas as pd
import numpy as np

print(f"Pandas Version  : {pd.__version__}")
print(f"Numpy Version   : {np.__version__}")

garis = "-"*70
```

```python
# load dataset

df = pd.read_csv("../data/synthetic_fooddelivery_dataset.csv")

display(df.head())
```

```python
# 1. Inspeksi struktur data


print("=== INFO DATAFRAME ===")
print(df.info())

print(garis)

# 2. Cek persentase missing values
print("=== PERSENTASE MISSING VALUES (%) ===")
missing_pct = (df.isna().mean() * 100).round(2)
print(missing_pct[missing_pct > 0].sort_values(ascending=False))

print(garis)

# 3. Cek duplikat data
print("=== JUMLAH BARIS DUPLIKAT ===")
print(df.duplicated().sum())
```

--- Sprint 2

```python
# copyan 'df = df_clean' agar aman

df_clean = df.copy()

# === task 1: mengubah tipe data kolom 'Waktu_Transaksi' ===

df_clean["Waktu_Transaksi"] = pd.to_datetime(
    df_clean["Waktu_Transaksi"], format="mixed"
)

# Ekstrak 'Waktu_Transaksi' menjadi jam dan hari
df_clean["Jam"] = df_clean["Waktu_Transaksi"].dt.hour
df_clean["Hari_Transaksi"] = df_clean["Waktu_Transaksi"].dt.day_name()


# === task 2: Handling Missing Values (Imputation) ===

# 1. 'Jarak_Kirim_KM' --> drop 7% baris yang jaraknya kosong
df_clean = df_clean.dropna(subset=["Jarak_Kirim_KM"])

# 2. 'Rating_Pelanggan' -  20% --> isi dengan median dari kategori menu masing"
df_clean["Rating_Pelanggan"] = df_clean.groupby("Kategori_Menu")[
    "Rating_Pelanggan"
].transform(lambda x: x.fillna(x.median()))

# 3. 'Ulasan_Teks' - 20% --> isi pakai str "Tidak ada ulasan"
df_clean['Ulasan_Teks'] = df_clean['Ulasan_Teks'].fillna("Tidak ada ulasan")


# === task 3: Verifikasi Hasil ===

print("=== SISA MISSING VALUES SETELAH CLEANING ===")
print(df_clean.isna().sum())

print(garis)

print("=== 5 BARIS PERTAMA SETELAH CLEANING ===")
display(df_clean.head())

print(garis)

print("=== TIPE DATA WAKTU TRANSAKSI ===")
print(df_clean['Waktu_Transaksi'].dtype)
```

--- Sprint 3

```python
# task 1: Feature Engineering (Kategori Waktu)

# membuat kolom baru 'Kategori_waktu' berdasarkan kolom 'Jam'
kondisi = [
    (df_clean['Jam'] >= 6) & (df_clean['Jam'] <= 11),
    (df_clean['Jam'] >= 12) & (df_clean['Jam'] <= 15),
    (df_clean['Jam'] >= 16) & (df_clean['Jam'] <= 21)
]

pilihan = ['Pagi', 'Siang', 'Malam']

df_clean['Kategori_Waktu'] = np.select(kondisi, pilihan, default='Dini Hari')


# task 2: Agregasi data (Groupby)

# rata-rata harga, median rating, dan total pesanan per kategori menu
menu_stats = df_clean.groupby('Kategori_Menu').agg(
    Rata_rata_Harga=('Harga_Pesanan', 'mean'),
    Median_Rating=('Rating_Pelanggan', 'median'),
    Total_Pesanan=('ID_Pesanan', 'count')
).reset_index() # reset_index biar Kategori_Menu jadi kolom biasa, bukan index

print("=== PERFORMA KATEGORI MENU ===")
display(menu_stats.sort_values(by='Total_Pesanan', ascending=False))

# task 3: Pivot Table (Analisis Waktu vs Promo)

# rata-rata waktu tunggu (menit)
# berdasarkan kategori waktu dan status promo
pivot_waktu = pd.pivot_table(
    df_clean,
    values='Waktu_Tunggu_Menit',
    index='Kategori_Waktu',
    columns='Status_Promo',
    aggfunc='mean'
)

print("\n=== RATA-RATA WAKTU TUNGGU (MENIT) BERDASARKAN WAKTU & PROMO ===")
display(pivot_waktu.round(2)) # dibulatkan 2 desimal biar rapi
```

--- Sprint 4

```python
# task 1: Handle Outlier Harga (Pakai IQR)

# kita hitung batas bawah dan batas atas yang wajar pakai numpy
Q1 = np.percentile(df_clean['Harga_Pesanan'], 25)
Q3 = np.percentile(df_clean['Harga_Pesanan'], 75)
IQR = Q3 - Q1

batas_bawah = Q1 - (1.5 * IQR)
batas_atas = Q3 + (1.5 * IQR)

print(f"Batas Bawah Harga Wajar: {batas_bawah}")
print(f"Batas Atas Harga Wajar: {batas_atas}")

print("-"*60, ">")

# Filter datanya: Hanya ambil harga yang di antara batas bawah & atas
df_final = df_clean[(df_clean['Harga_Pesanan'] >= batas_bawah) & (df_clean['Harga_Pesanan'] <= batas_atas)]

print(f"Jumlah baris sebelum filter: {len(df_clean)}")
print(f"Jumlah baris setelah filter: {len(df_final)}")

# task 2: Korelasi (Hubungan antar variabel)

# Apakah jarak yang jauh bikin waktu tunggu lama? Apakah waktu tunggu lama bikin rating jelek?
kolom_numerik = ['Jarak_Kirim_KM', 'Waktu_Tunggu_Menit', 'Rating_Pelanggan', 'Harga_Pesanan']
korelasi = df_clean[kolom_numerik].corr()

print("\n=== MATRIKS KORELASI ===")
display(korelasi.round(2))

# task 3: Top 5 Pesanan Termahal (Setelah Outlier Dibuang)
print("\n=== 5 PESANAN TERMAHAL (DATA BERSIH) ===")
top_5 = df_final.nlargest(5, 'Harga_Pesanan')[['ID_Pesanan', 'Kategori_Menu', 'Harga_Pesanan', 'Jarak_Kirim_KM']]
display(top_5)
```
